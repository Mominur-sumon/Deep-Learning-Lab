## Train Yolov8

```python
from ultralytics import YOLO

# Load a pretrained YOLOv8 model
model = YOLO('yolov8n.pt')

# Train on your dataset
model.train(
    data='/content/widerface_yolo_format/dataset.yaml',
    epochs=20,
    imgsz=640,
    batch=16,
    name='widerface_yolov8'
)
```

## Check Score

```python
# Load the best trained model
model = YOLO('/content/runs/detect/widerface_yolov8/weights/best.pt')

# Evaluate on the validation set
metrics = model.val(
    data='/content/widerface_yolo_format/dataset.yaml',
    imgsz=640,
    split='val'
)

# print metrics
print(metrics)
```

## 6. Predict on Image

## 9. FPS Checking

```python
import time
import os
from ultralytics import YOLO

# Load model
model = YOLO('/content/runs/detect/widerface_yolov8/weights/best.pt')

# Get image paths
image_folder = '/content/test_sample'
image_list = sorted(os.listdir(image_folder))
image_paths = [os.path.join(image_folder, img) for img in image_list]

# Time the prediction
start = time.time()

results = model.predict(
    source=image_paths,
    stream=False,     # safe for 50 images
    save=False,
    verbose=False
)

end = time.time()

# Compute FPS
total_images = len(image_paths)
total_time = end - start
fps = total_images / total_time

# Display results
print(f'🖼️ Total images: {total_images}')
print(f'⏱️ Total time: {total_time:.2f} seconds')
print(f'⚡ Average FPS: {fps:.2f}')
```
