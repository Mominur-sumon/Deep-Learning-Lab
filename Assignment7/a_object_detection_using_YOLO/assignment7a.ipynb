## YOLOv8, YOLOv11 and YOLOv12 models and object detection capability, speed.

```python
import cv2
import time
from ultralytics import YOLO
import pandas as pd
import os

def evaluate_model(model_path, video_path, output_name):
    model = YOLO(model_path)
    cap = cv2.VideoCapture(video_path)

    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    out = cv2.VideoWriter(output_name, cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

    total_frames = 0
    total_objects = 0
    total_fps = 0

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        start = time.time()
        results = model(frame)
        end = time.time()

        frame_fps = 1 / (end - start)
        total_fps += frame_fps
        total_frames += 1
        total_objects += len(results[0].boxes)

        # print(f"Processing frame {total_frames}: Detected {len(results[0].boxes)} objects at {frame_fps:.2f} FPS")
        annotated = results[0].plot()
        out.write(annotated)

    cap.release()
    out.release()

    avg_fps = total_fps / total_frames if total_frames else 0

    return {
        "Model": os.path.basename(model_path),
        "FPS": round(avg_fps, 2),
        "Objects Detected": total_objects,
        "Frames": total_frames,
        "Output File": output_name
    }


video_path = "./datasets1.mp4"

results = []
results.append(evaluate_model("yolov8n.pt", video_path, "output_v8.mp4"))
results.append(evaluate_model("yolo11m.pt", video_path, "output_v11.mp4"))
results.append(evaluate_model("yolo12n.pt", video_path, "output_v12.mp4"))

# comparison table
df = pd.DataFrame(results)
print("\n YOLO Model Comparison:")
print(df.to_string(index=False))
```
